# Linear Regression Example

This notebook demonstrates a simple linear regression example based on the provided document pages. We will use the advertising dataset to predict sales based on TV advertising spending.

```python
# **Step 1: Import Libraries and Suppress Warnings**

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn import metrics

import warnings
warnings.filterwarnings('ignore')
```

# **Step 2: Load and Explore the Data**

```python
# **Read the data**

advertising = pd.read_csv('advertising.csv')

# **Display the first few rows**

advertising.head()
```

```python
# **Get data information**

advertising.info()
```

```python
# **Check the shape**

print("Data shape =", advertising.shape)
```

# **Step 3: Data Cleaning and Visualization**

```python
# **Check for missing values**

print('Checking for Null values')
print(advertising.isnull().sum()*100/advertising.shape[0])
```

```python
# **Outlier Analysis**

fig, axs = plt.subplots(3, figsize = (5,5))
plt1 = sns.boxplot(advertising['TV'], ax = axs[0])
plt2 = sns.boxplot(advertising['Newspaper'], ax = axs[1])
plt3 = sns.boxplot(advertising['Radio'], ax = axs[2])
plt.tight_layout()
plt.show()
```

```python
# **Visualize relationships between features and target**

sns.pairplot(advertising, x_vars=['TV', 'Newspaper', 'Radio'], y_vars='Sales', height=4, aspect=0.7, kind='scatter')
plt.show()
```

# **Step 4: Prepare the Data for Linear Regression**

```python
# **Create X (independent) and y (dependent) variables**

X = advertising['TV']
y = advertising['Sales']
```

```python
# **Split data into training and testing sets**

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, test_size=0.3, random_state=100)
```

```python
# **Add a constant to get an intercept**

X_train_sm = sm.add_constant(X_train)
```

# **Step 5: Build and Train the Linear Regression Model**

```python
# **Fit the resgression line using 'OLS'**

lr = sm.OLS(y_train, X_train_sm).fit()
```

```python
# **Print the parameters, i.e. intercept and slope of the regression line obtained**

print(lr.params)
```

```python
# **Performing a summary operation lists out all the different parameters of the regression line fitted**

print(lr.summary())
```

# **Step 6: Visualize the Model Fit**

```python
# **Visualize the regression line**

plt.scatter(X_train, y_train)
plt.plot(X_train, 6.948 + 0.054*X_train, 'r')
plt.show()
```

# **Step 7: Make Predictions on the Test Set**

```python
# **Add a constant to X_test**

X_test_sm = sm.add_constant(X_test)
```

```python
# **Predict the y values corresponding to X_test_sm**

y_pred = lr.predict(X_test_sm)
```

# **Step 8: Evaluate the Model**

```python
# **Compute model evaluation metrics**

print("RMSE :", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print("R-squared :", metrics.r2_score(y_test, y_pred))
```

```python
# **Visualize the fit on the test set**

plt.scatter(X_test, y_test)
plt.plot(X_test, 6.948 + 0.054 * X_test, 'r')
plt.show()
```
